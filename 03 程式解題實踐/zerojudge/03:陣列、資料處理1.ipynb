````python
#a015. 矩陣的翻轉
while True: #測資有多筆資料
  try:
    r,s=map(int,input().split())
    max=[]
    for i in range(r):
      row=list(map(int,input().split())) #沒有括號 .split()要記得括號
      max.append(row)
    for j in range(s):
      for i in range(r):
        print(max[i][j],end=" ")  #end=" "防止換行
        #先固定j(原本的列)，然後再跑i(原本的行)，
        #就可以讓輸出的順序變成垂直讀取，達成 90 度翻轉的效果。
      print()
  except EOFError:
    break
````

````python
#a417 螺旋矩陣
t=int(input())
for _ in range(t):
  n,m=map(int,input().split())
  s=[[0]*n for _ in range(n)]
  r,c=0,0
  if m==1:  #右下左上
    dr=[0,1,0,-1]
    dc=[1,0,-1,0]
  else:    #下右上左
    dr=[1,0,-1,0]
    dc=[0,1,0,-1]
  d=0  #方向
  for i in range(1,n*n+1):
    s[r][c]=i
    nr,nc=r+dr[d],c+dc[d]  #下一步
    if 0 <= nr < n and 0 <= nc < n and s[nr][nc] == 0:
      r,c=nr,nc
    else:
      d=(d+1)%4
      r,c=r+dr[d],c+dc[d] #轉向並變往下一格
  for j in s:
    print("".join(f"{x:5d}"for x in j)) #加""防止格式跑掉
````

````python
#d904 換零錢
c,n=map(int,input().split())
s=[]
for _ in range(n):
  s.append(int(input()))
dp=[int(10000000)]*(c+1)
dp[0]=0
for i in range(1,c+1):
  for coin in s:
    if i-coin>=0:
      dp[i]=min(dp[i],dp[i-coin]+1)
print(dp[c])
````

```python
c,n=map(int,input().split())
s=[]
for _ in range(n):
  s.append(int(input()))
s.sort(reverse=True)
times=0
while c>0:
  for i in s:
    if c-i>=0:
      times+=1
      c-=i
print(times)
```
這是我一開始的程式碼，想說從最大額的硬幣開始找，但是沒考慮到特殊情況  
也就是當c是幣值的倍數時可能出現問題   
Ex:c=6 ,s=[4,3,1] , 用我上面的方法得出times=3 (4,1,1),然而答案是2 (3,3)  
後來查詢答案找到了動態規劃這個解法(dp)   
這方法將大問題拆解成小問題，並透過儲存已解決的答案來避免重複運算



````python
#e357. 遞迴函數練習
def F(x):
  if x==1:
    return(1)
  elif x%2==0:
    return F(x/2)
  else:
    return F(x-1)+F(x+1)
x=int(input())
print(F(x))
````

````python
#h664 河內塔
t=int(input())
for _ in range(t):
  n,s=map(int,input().split())
  if s%2==1:
    print(1)
  else:
    c=0
    while s%2==0:
      s//=2
      c+=1
    print(c+1)
````

如果s(步數)是奇數，則移動圓盤1  
如果s是偶數，要看s可以被2除幾次，最後在加1  
找規律，圓盤 1 每隔一步就會出現一次,圓盤 2 每隔四步會出現一次,圓盤 3 每隔八步會出現一次
圓盤 i 出現的步數，剛好都是 2^(i-1) 的奇數倍    
因為次數減1，所以最後加1就是答案
