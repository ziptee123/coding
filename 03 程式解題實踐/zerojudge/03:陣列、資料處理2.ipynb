````python
#c081. 00102 - Ecological Bin Packing
while True:
  try:
    bgc=list(map(int,input().split()))
    sol=[("BCG",[0, 5, 7]),  #文字和顏色排序對應
      ("BGC",[0, 4, 8]),
      ("CBG",[2, 3, 7]),
      ("CGB",[2, 4, 6]),
      ("GBC",[1, 3, 8]),
      ("GCB",[1, 5, 6])]
    total=sum(bgc)
    mintime=float("inf")
    word=""
    for i,j in sol:
      time=total-sum(bgc[k] for k in j) #總和減保留不搬的數量即為搬動的次數
      if time < mintime:
        mintime = time  #取最小值
        word = i
    print(word,mintime)
  except EOFError:
    break
````

````python
#s140 排隊買飲料
n, a, b = map(int, input().split())
low = 0
high = min(a, b) * n
ans = high
while low <= high:
  mid = (low + high) // 2
  #在 mid 秒時，兩台機器總共產出的飲料數量
  if (mid // a) + (mid // b) >= n:
    ans = mid        #夠喝，紀錄時間,縮短（往左找）
    high = mid - 1
  else:
    low = mid + 1    #不夠喝，增加時間（往右找）
print(ans % ((10**6)+7))
````
