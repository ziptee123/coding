````python
#a005. Eva 的回家作業
t =int(input()) #幾次
for i in range(t):
  num1,num2,num3,num4=map(int,input().split())
  #先判斷等差，剩下情況都是等比
  if (num2 - num1) == (num3 - num2):
    print(f"{num1} {num2} {num3} {num4} {(num3-num2)+num4}")
  else:
    det=num2//num1 #等比，使用整數除法，不然會有小數點
    print(f"{num1} {num2} {num3} {num4} {det*num4}")
````

````python
#a034. 二進位制轉換
while True:
  try:
    n=int(input())
    if n==0:
      print(0)
    else:
      res=""
      while n>0: #連除
        rem=n%2 #餘數進位
        res=str(rem)+res #用字串形式往後加
        n=n//2
      print(res)
  except EOFError:
    break
````

````python
#a095 麥哲倫的陰謀
while True:
  try:
    n,m=map(int,input().split())
    if n==m:
      print(m)
    else:
      print(m+1)
  except EOFError:
    break
````

若有兩頂以上的紅帽子，那麼大家第一天都不敢出去。  

到了第二天，因為第一天沒人敢出去，大家都會知道至少有兩頂紅帽子。

如果只看到一頂帽子的人就會知道自己是紅帽而出獄。(紅帽只有兩頂)

那如果第二天也沒人敢出獄，第三天大家就知道昨天每個人都至少看到了兩頂紅帽子。

也就是說，大家就能確定紅帽至少有三頂，以此類推。

所以有M頂紅帽，到了第M天就會知道至少有M頂紅帽。

所以只看到M-1頂紅帽的人(也就是自己戴著一頂紅帽的人)就可以出獄。

到了隔天，看到紅帽出獄的白帽們就敢出獄了。

所以如果M=N(大家都是紅帽)，大家第M天就會一起出去。

如果M!=N，白帽就要等到M+1天才能出獄。

````python
#a148. You Cannot Pass?!
while True:
  try:
    num=list(map(int,input().split()))
    n=num[0]
    del(num[0]) #刪除不是分數的
    s=sum(num)/n #計算平均
    if s>59: #及格分>60
      print("no")
    else:
      print("yes")
  except EOFError:
    break
````

````python
#a271 彩色蘿蔔
n=int(input())
for _ in range(n):
  x,y,z,w,n,m=map(int,input().split())
  s=list(map(int,input().split()))
  poison=0
  for i in s:
    m-=n*poison  #要先扣血，不能吃完才扣
    if m>0:
      if i==1:
        m+=x
      elif i==2:
        m+=y
      elif i==3:
        m-=z
      elif i==4:
        m-=w
        poison+=1
  if m>0:
    print(f"{m}g")
  else:
    print("bye~Rabbit")
````

````python
#d058. BASIC 的 SGN 函數(不可使用if)
#本題運用了python中布林值參與計算時會變成0,1的特性
n=int(input())
ans=(n>0)-(n<0)
#當n>0時,ans=True-False=1-0=1
#當n=0時,ans=False-False=0-0=0
#當n<0時,ans=False-True=0-1=-1
print(ans)
````

````python
#d066. 上學去吧！
hh, mm = map(int, input().split())
if (hh == 7 and mm >= 30) or 7 < hh < 17: #在校時間
    print("At School")
else: #剩下的就是Off School
    print("Off School")
````
