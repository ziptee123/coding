````python
#a004. 文文的求婚
while True:
  try:
    y = int(input())
    output="平年"
    if y%4==0:
      output="閏年"
    if y%100==0:
      output="平年"
    if y%400==0:
      output="閏年"
    print(output)
  except EOFError:
    break
````

````python
#a015. 矩陣的翻轉
while True: #測資有多筆資料
  try:
    r,s=map(int,input().split())
    max=[]
    for i in range(r):
      row=list(map(int,input().split())) #沒有括號 .split()要記得括號
      max.append(row)
    for j in range(s):
      for i in range(r):
        print(max[i][j],end=" ")  #end=" "防止換行
        #先固定j(原本的列)，然後再跑i(原本的行)，
        #就可以讓輸出的順序變成垂直讀取，達成 90 度翻轉的效果。
      print()
  except EOFError:
    break
````

````python
#a104. 排序
while True:
  try:
    n=int(input())
    num=list(map(int,input().split()))
    num.sort() #.sort()預設由小排到大
    print(*num) #星號解包
  except:
    break
````

````python
#c013. 00488 - Triangle Wave
n=int(input())
ny=input() #留一個input()接收測資空白換行
for i in range(n):
  a=int(input())
  f=int(input())
  for j in range(f):
    for k in range(1,a+1): #注意a+1不是a
      print(str(k)*k) #改成字串才能疊加
    for k in range(a-1,0,-1): #倒著回去
      print(str(k)*k)
    print("") #換行
````

````python
#d072(d069) 格瑞哥里的煩惱 (Case 版)
n=int(input())
for i in range(n):
  y=int(input())
  if y%100==0:
    if y%400==0:
      print(f"Case {i+1}: a leap year")
    else:
      print(f"Case {i+1}: a normal year")
  elif y%4==0:
    print(f"Case {i+1}: a leap year")
  else:
    print(f"Case {i+1}: a normal year")
````

````python
#d904 換零錢
c,n=map(int,input().split())
s=[]
for _ in range(n):
  s.append(int(input()))
dp=[int(10000000)]*(c+1)
dp[0]=0
for i in range(1,c+1):
  for coin in s:
    if i-coin>=0:
      dp[i]=min(dp[i],dp[i-coin]+1)
print(dp[c])
````

d904 換零錢   
```python
c,n=map(int,input().split())
s=[]
for _ in range(n):
  s.append(int(input()))
s.sort(reverse=True)
times=0
while c>0:
  for i in s:
    if c-i>=0:
      times+=1
      c-=i
print(times)
```
這是我一開始的程式碼，想說從最大額的硬幣開始找，但是沒考慮到特殊情況  
也就是當c是幣值的倍數時可能出現問題   
Ex:c=6 ,s=[4,3,1] , 用我上面的方法得出times=3 (4,1,1),然而答案是2 (3,3)  
後來查詢答案找到了動態規劃這個解法(dp)   
這方法將大問題拆解成小問題，並透過儲存已解決的答案來避免重複運算



````python
#e357. 遞迴函數練習
def F(x):
  if x==1:
    return(1)
  elif x%2==0:
    return F(x/2)
  else:
    return F(x-1)+F(x+1)
x=int(input())
print(F(x))
````
