```python
#物件導向
class car: #class(類別)
  wheels=4
  def __init__(self,make,color): #初始化,self是自己
    self.make=make
    self.color=color
  def turn_on(self):
    print(f"{self.make} {self.color} is turning on")
    return self
  def drive(self):
    print(f"{self.make} {self.color} is driving")
    return self
Car=car("Toyota","blue")
print(Car.make,Car.color,Car.wheels)
class truck(car): #繼承
  def __init__(self,make,color,weight):
    super().__init__(make,color) #super()可以使用父類別
    self.weight=weight
  def stop(self):
    print(f"{self.make} {self.color} stopped")
Truck=truck("Benz","red","2T")
Truck.stop()
Truck.drive()
Car.turn_on().drive() #用return讓Car.turn_on()還是car
print(Truck.weight)
class bicycle:
  def drive(self): #使用相同的函式名稱一樣可以
    print(f"driving...")
def start(vehicle): #通用的函式
  vehicle.drive()
bike=bicycle()
start(Car)
start(bike)
```
